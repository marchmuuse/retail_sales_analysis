```python
query = """
select categories.item_category_name,
sum(sales.item_price*sales.item_cnt_day) as revenue,
sum(sales.item_cnt_day) as units_sold
from sales
inner join items
on sales.item_id = items.item_id
inner join categories
on items.item_category_id = categories.item_category_id
group by categories.item_category_name
order by revenue desc
limit 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
select categories.item_category_name,
sum(sales.item_price*sales.item_cnt_day) as revenue,
sum(sales.item_cnt_day) as units_sold,
sum(sales.item_price*sales.item_cnt_day) / sum(sales.item_cnt_day) as avg_price
from sales
inner join items
on sales.item_id = items.item_id
inner join categories
on items.item_category_id = categories.item_category_id
group by categories.item_category_name
order by revenue desc
limit 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
select items.item_name,
categories.item_category_name,
sum(sales.item_price*sales.item_cnt_day) as revenue,
sum (item_cnt_day) as units_sold
from sales
inner join items
on sales.item_id = items.item_id
inner join categories
on items.item_category_id = categories.item_category_id
group by items.item_id, items.item_name, categories.item_category_name
order by revenue desc
limit 10;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
select date_block_num,
sum(item_price*item_cnt_day) as revenue
from sales
group by date_block_num
order by date_block_num;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
with monthly as (
select date_block_num,
sum(item_price*item_cnt_day) as revenue
from sales
group by date_block_num
)
select date_block_num,
revenue,
lag(revenue) over (order by date_block_num) as previous_revenue
from monthly
order by date_block_num;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
with monthly as (
select date_block_num,
sum(item_price*item_cnt_day) as revenue
from sales
group by date_block_num
),
monthly_with_previous as (
select date_block_num,
revenue,
lag(revenue) over (order by date_block_num) as previous_revenue
from monthly
)
select
date_block_num,
revenue,
previous_revenue,
(revenue - previous_revenue) / previous_revenue * 100 as revenue_growth_pct
from monthly_with_previous
order by date_block_num;
"""

pd.read_sql_query(query, conn)
```

```python
query = """
select
substr(date, 4, 2) as month,
sum(item_price*item_cnt_day) as revenue
from sales
group by month
order by month;
"""
pd.read_sql_query(query, conn)
```

```python
query = """
WITH shop_year AS (
    SELECT
        shop_id,
        substr(date, 7, 4) AS year,
        SUM(item_price * item_cnt_day) AS revenue
    FROM sales
    GROUP BY shop_id, year
)
SELECT
    shop_id,
    year,
    revenue,
    LAG(revenue) OVER (
        PARTITION BY shop_id
        ORDER BY year
    ) AS previous_year_revenue
FROM shop_year
ORDER BY shop_id, year;
"""
pd.read_sql_query(query, conn)
```

```python
import pandas as pd

sales = pd.read_csv("../data/sales_train.csv")

sales_clean = sales.drop_duplicates().copy()

sales_clean['date'] = pd.to_datetime(
    sales_clean['date'],
    format='%d.%m.%Y'
)

sales_clean = sales_clean[sales_clean['item_price'] != -1].copy()

sales_clean['revenue'] = (
    sales_clean['item_price'] * sales_clean['item_cnt_day']
)

print(sales_clean.shape)
```

```python
import matplotlib.pyplot as plt

monthly_revenue = (
    sales_clean
    .groupby(sales_clean['date'].dt.to_period('M'))['revenue']
    .sum()
)

monthly_revenue.index = monthly_revenue.index.to_timestamp()

plt.figure(figsize=(12, 5))
plt.plot(monthly_revenue.index, monthly_revenue.values)

plt.title('Динамика месячной выручки')
plt.xlabel('Дата')
plt.ylabel('Выручка, ₽')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
shop_revenue = (
    sales_clean
    .groupby('shop_id')['revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10, 6))
plt.barh(
    shop_revenue.index.astype(str),
    shop_revenue.values
)

plt.title('Топ-10 магазинов по выручке')
plt.xlabel('Выручка, ₽')
plt.ylabel('Магазин')
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
items = pd.read_csv("../data/items.csv")

print(items.shape)
print(items.columns)
```

```python
product_revenue = (
    sales_clean
    .groupby('item_id')['revenue']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

product_revenue = product_revenue.merge(
    items[['item_id', 'item_name']],
    on='item_id',
    how='left'
)

plt.figure(figsize=(12, 6))
plt.barh(
    product_revenue['item_name'],
    product_revenue['revenue']
)

plt.title('Топ-10 товаров по выручке')
plt.xlabel('Выручка, ₽')
plt.ylabel('Товар')
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
product_units = (
    sales_clean
    .groupby('item_id')['item_cnt_day']
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)

product_units = product_units.merge(
    items[['item_id', 'item_name']],
    on='item_id',
    how='left'
)

plt.figure(figsize=(12, 6))
plt.barh(
    product_units['item_name'],
    product_units['item_cnt_day']
)

plt.title('Топ-10 товаров по количеству проданных единиц')
plt.xlabel('Количество единиц')
plt.ylabel('Товар')
plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```

```python
monthly_avg_price = (
    sales_clean
    .groupby(sales_clean['date'].dt.to_period('M'))
    .apply(lambda x: x['revenue'].sum() / x['item_cnt_day'].sum())
)

monthly_avg_price.index = monthly_avg_price.index.to_timestamp()

plt.figure(figsize=(12, 5))
plt.plot(monthly_avg_price.index, monthly_avg_price.values)

plt.title('Динамика средней цены продажи')
plt.xlabel('Дата')
plt.ylabel('Средняя цена, ₽')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```
